# Problem Set
SD Retailing Inc. periodically monitors its warehouse management to generate key insights for its business. Warehouse management tools usually store and update information about available products in the form of inventory. SD Retailing Inc. tracks inventory in the following product categories: Hand soap, Moisturizer, Sunscreen, Body oil, and Shampoo. For each product in these categories, it monitors unit price (the price of each item), available quantity, expiration date of those products, and advance orders. Advance orders are orders customers place in advance. Expiration date tracks the “good days” left: 0 means it has expired. Once it has expired, the system removes the item, and employees also manually remove it from the warehouse so they do not have the item in physical form or in the digital system.

The code originally provided contained several issues. Some issues prevented the program from running, while others made the generated data inconsistent with the business context.

```python
# Import required library
import random
```

### Part A:
1. Correct all problems in the code.  
2. Run the corrected function to generate the inventory.  
3. Verify and print:  
• the total number of products generated,  
• the number of products in each product category,  
• the minimum and maximum expiration days appearing anywhere in the inventory.  
4. Print the complete information for the first product in each category.

```python
# Set the random seed for reproducibility
random.seed(42)

# Correct the issues with the code and make it consistent with the business context in the instructions text
def generate_inventory(num_categories=5, num_products_per_category=10):
    '''This user-defined function generates a hypothetical inventory for SD Retailing.'''

    # Create a categories list and instantiate an inventory dictionary
    categories = ['Hand soap', 'Moisturizer', 'Sunscreen', 'Body oil', 'Shampoo']
    inventory = {}

    # Generate the SD Retailing Inc. inventory
    for category in categories:
        inventory[category] = []
        for _ in range(num_products_per_category):
            # Update the `Expiration Date` portion of the code be to be > 0 based on business rules
            product = {
                'Quantity': random.randint(10, 100),
                'Unit Price': round(random.uniform(5.0, 500.0), 2),
                'Expiration Date': random.randint(1, 365),
                'Advanced Ordered': random.randint(0, 8)
            }
            inventory[category].append(product)
    return inventory

# Run the corrected function and assign the inventory results to a new variable
inventory_data = generate_inventory()

# Initialize the total products variable and instantiate an expirations list to verify and print statistics
total_products = 0
all_expirations = []

# Generate category statistics and print number of products in each category
# Uses `.items()` to simultaneously access keys and values, which is more 
# efficient than separate lookups and avoids needing to merge data later
print('*** Products Per Category ***')
for category, products in inventory_data.items():
    count = len(products)
    total_products += count
    print(f'{category}: {count} products')
    for p in products:
        all_expirations.append(p['Expiration Date'])

# Print the total number of products generated
print(f'\nTotal products generated: {total_products}')

# Print min and max expiration days within entire inventory
print(f'\nMin. Expiration Days: {min(all_expirations)}')
print(f'Max. Expiration Days: {max(all_expirations)}')

# Using .items() here again ensures clean, readable access to both category names 
# and their corresponding product data in a single iteration.
print('\n*** First Product in Each Category ***')
for category, products in inventory_data.items():
    print(f'{category}: {products[0]}')
```

### Part B: Identifying High-Impact Near-Expiration Inventory
SD Retailing wants to identify products that represent relatively high financial exposure and require attention because they are approaching expiration.  
  
A product is considered high-impact near-expiration inventory when all three of the following conditions are satisfied:  
* Its Unit Price is greater than $200.
* It has fewer than 30 good days left before expiration.
* Its Available Quantity is greater than its Advanced Ordered quantity.

Using Python, complete the following tasks:
1. Identify all qualifying products that satisfy the three conditions above.
2. For each qualifying product, print the following information:  
    o Product Category  
    o Quantity  
    o Unit Price  
    o Expiration Date  
    o Advanced Ordered  
    o Inventory Value  
   
    Calculate Inventory Value as:
  
    Inventory Value = Quantity × Unit Price  
   
4. Print the total number of qualifying products.  
5. Calculate and print the aggregate inventory value of all qualifying products.  
6. Calculate and print the following ratio:  
  
    Average Quantity ÷ Average Advanced Ordered
   
    If the average Advanced Ordered quantity is zero, your program should handle this
    situation appropriately rather than producing an error.
     
7. Identify and print the product category that contributes the largest dollar amount to
   the aggregate inventory value of the qualifying products.  
  
**Important**: Do not manually determine the answers by inspecting the printed data. Our Python 
program must perform the required calculations and determine the results.

```python
# Initialize variables and instantiate lists/dictionary for calculating required data/statistics
qualifying_product_count = 0
total_qualifying_value = 0.0
qualifying_quantities = []
qualifying_advanced_orders = []
category_values = {} # To track value per category for top contributing category

# Identify qualifying products and print their info
print('*** High-Impact Near-Expiration Inventory ***')
print('- ' * 14)
for category, products in inventory_data.items():
    for product in products:
        # Check if the three conditions defined in Part B are all met
        if (product['Unit Price'] > 200 and
            product['Expiration Date'] < 30 and
            product['Quantity'] > product['Advanced Ordered']):

            # Increment the count of qualifying products
            qualifying_product_count += 1
            # Calculate the inventory value for each qualifying product
            inventory_value = product['Quantity'] * product['Unit Price']
            # Add the inventory value to the total
            total_qualifying_value += inventory_value

            # Store data for averages and category tracking
            qualifying_quantities.append(product['Quantity'])
            qualifying_advanced_orders.append(product['Advanced Ordered'])
            category_values[category] = category_values.get(category, 0) + inventory_value

            # Print individual product details
            # Use `:.2f` instead of `round()` for display because it maintains 
            # trailing zeros and the comma adds thousand separators for readability
            print(f'Product Category: {category}\n' 
                  f'Quantity: {product['Quantity']}\n'
                  f'Unit Price: ${product['Unit Price']}\n'
                  f'Expiration Date: {product['Expiration Date']} days\n'
                  f'Advanced Ordered: {product['Advanced Ordered']}\n'
                  f'Inventory Value: ${inventory_value:,.2f}')
            print('- ' * 14)

print('\n*** Qualifying Products Data ***')
# Print total number of qualifying products
print(f'Number of qualifying products: {qualifying_product_count}')

# Print aggregate inventory value, formatted to display trailing zeros and comma for thousand separator
print(f'Aggregate Inventory Value: ${total_qualifying_value:,.2f}')

# Calculate and print ratio
# Use inline 'if-else' to return 0 if the list is empty, preventing a ZeroDivisionError if no products qualify
avg_qty = sum(qualifying_quantities) / len(qualifying_quantities) if qualifying_quantities else 0
avg_adv = sum(qualifying_advanced_orders) / len(qualifying_advanced_orders) if qualifying_advanced_orders else 0

# Print the required ratio while handling possible division by zero
if avg_adv == 0:
    print('Ratio (Avg Qty / Avg Adv Ordered): N/A (Average Advanced Ordered is zero)')
else:
    print(f'Ratio (Avg Qty / Avg Adv Ordered): {avg_qty / avg_adv:.2f}')

# Identify the top contributing category
# `max()` iterates keys but using `.get` forces it to compare their values, returning the key with the highest dollar amount
top_category = max(category_values, key=category_values.get)
print(f'Category with largest dollar impact: {top_category} (${category_values[top_category]:,.2f})')
```

### Part C:
Management now wants to compare prices among the products identified in Part B.  

Using only the qualifying products from Part B, normalize each qualifying product's Unit Price by identifying the highest unit price and dividing every other product’s unit price by that number
1. Store the normalized value as a new piece of information for each qualifying product. Do not replace or modify the original Unit Price.
2. Print:  
• the original Unit Price of the most expensive qualifying product,  
• the original Unit Price of the least expensive qualifying product,  
• the normalized Unit Price of the least expensive qualifying product.     
3. Print all qualifying products in order from the highest normalized Unit Price to the lowest.  

At the end of Part C, write 2–3 sentences explaining what a normalized Unit Price close to 1 means in this business context.

```python
# Identify qualifying products again, assigning to new lists and variables for Part C statistics

# Instantiate a new list to hold the qualifying products data (list of dictionaries)
qualifying_products = []

# Identify (again) qualifying products and their info for future calculations and printing
for category, products in inventory_data.items():
    for product in products:
        if (product['Unit Price'] > 200 and
            product['Expiration Date'] < 30 and
            product['Quantity'] > product['Advanced Ordered']):
            # Create a copy to store values without changing original `inventory_data`
            prod_info = product.copy()
            prod_info['Category'] = category
            qualifying_products.append(prod_info)

# Define a helper function to avoid using lambda or external libraries to use normalized price when sorting our list later
def get_normalized_price(product):
    return product['Normalized Price']

# Proceed only if qualifying products exist
if qualifying_products:
    # Identify highest and lowest prices
    highest_price = max(p['Unit Price'] for p in qualifying_products)
    lowest_price = min(p['Unit Price'] for p in qualifying_products)

    # Normalize each product's unit price and store as new info
    for p in qualifying_products:
        p['Normalized Price'] = p['Unit Price'] / highest_price

    # Print required price statistics
    print(f'Original Unit Price (Most Expensive): ${highest_price:,.2f}')
    print(f'Original Unit Price (Least Expensive): ${lowest_price:,.2f}')
    print(f'Normalized Unit Price (Least Expensive): {(lowest_price / highest_price):.2f}')

    # Print all qualifying products sorted from highest normalized price to lowest
    print('\n*** Qualifying Products (Sorted by Normalized Price) ***')
    print('- ' * 15)

    # Generate a new list using `sorted()`; the 'key' argument uses our helper function 
    # to extract the 'Normalized Price' for comparison, while 'reverse=True' 
    # ensures the items are arranged in descending order (highest to lowest)
    sorted_list = sorted(qualifying_products, key=get_normalized_price, reverse=True)

    for p in sorted_list:
        print(f'Product Category: {p['Category']}\n'
              f'Normalized Price: {p['Normalized Price']:.2f}\n'
              f'Original Unit Price: ${p['Unit Price']:,.2f}')
        print('- ' * 15)

print('\n*** Business Context Explanation ***' 
      '\nA normalized Unit Price close to 1 means the product\'s price is at or very near the maximum price found in this high-risk group.'
      '\nIn this context, it indicates items that pose the highest individual financial loss per unit if they expire before being sold.')
```

### Part D:
Verify whether removing the single most expensive qualifying product changes the category that contributes to the largest aggregate Inventory Value.

```python
# Define a helper function to extract price for identification of the most expensive product
def get_unit_price(product):
    return product['Unit Price']

# Identify the most expensive qualifying product using the helper function
most_expensive_product = max(qualifying_products, key=get_unit_price)

# Create a new list excluding only that specific most expensive product instance
# Reusing `qualifying_products` from Part C (using a list comprehension instead of 4 single lines of code)
remaining_products = [p for p in qualifying_products if p is not most_expensive_product]

# Instantiate a new dictionary to track value per category for the top 
# contributing category after removing the most expensive qualifying product 
new_category_values = {}

# Calculate aggregate values per category for the remaining products
for p in remaining_products:
    cat = p['Category']
    new_category_values[cat] = new_category_values.get(cat, 0) + (p['Quantity'] * p['Unit Price'])

# Identify the new top category and compare with `top_category` from Part B
new_top_category = max(new_category_values, key=new_category_values.get)

# Print which product was removed and how it affected the results
print(f'Removed: {most_expensive_product['Category']} category product at ${most_expensive_product['Unit Price']:,.2f}')
print(f'Previous Category with largest impact: {top_category}')
print(f'New Category with largest impact: {new_top_category} (${new_category_values[new_top_category]:,.2f})')

# Check to see if the new category that contributes the largest dollar amount to the aggregate inventory 
# value of the qualifying products is the same as the the previous category that was the top contributor
if new_top_category == top_category:
    print('\nResult: The top contributing category remained the same.')
else:
    print('\nResult: The top contributing category has changed!')
```
